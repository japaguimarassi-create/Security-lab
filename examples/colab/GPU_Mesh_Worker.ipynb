# GPU Mesh · Worker para Google Colab

Este notebook conecta a sessão Colab atual ao GPU Mesh. Selecione uma GPU/TPU nas configurações do Colab antes de executar. O runtime é temporário: quando a sessão termina, o worker fica offline. O compartilhamento de recursos começa desativado.

```python
%pip install --upgrade "klod-ai[worker] @ git+https://github.com/japaguimarassi-create/Security-lab.git"
```

```python
import getpass
import os

os.environ['GPU_MESH_API_URL'] = input('URL HTTPS da API GPU Mesh: ').strip()
os.environ['GPU_MESH_ACCESS_TOKEN'] = getpass.getpass('Token de sessão da conta GPU Mesh: ').strip()
os.environ['GPU_MESH_WORKER_STATE'] = '/content/.gpu-mesh/colab-worker.json'
os.environ['GPU_MESH_HEARTBEAT_SECONDS'] = '30'
os.environ['GPU_MESH_CPU_FRACTION'] = '0.5'
os.environ['GPU_MESH_RAM_FRACTION'] = '0.5'
share = input('Compartilhar a capacidade desta sessão? Digite SHARE para permitir: ').strip()
os.environ['GPU_MESH_SHARE'] = 'true' if share == 'SHARE' else 'false'
assert os.environ['GPU_MESH_API_URL'].startswith('https://')
assert os.environ['GPU_MESH_ACCESS_TOKEN']
os.environ['GPU_MESH_MAX_TASK_SECONDS'] = '3600'
print('Configuração carregada. Compartilhamento:', os.environ['GPU_MESH_SHARE'])
```

## Iniciar o worker

A próxima célula permanece ativa enquanto o agente procura tarefas. Não feche a sessão se quiser manter o worker disponível. A disponibilidade de GPU, a duração da sessão e os limites de uso continuam sujeitos às regras e cotas do Colab.

## Verificar hardware local (opcional)

Execute esta célula antes de iniciar o worker para conferir a GPU detectada.

```python
import torch
print('CUDA disponível:', torch.cuda.is_available())
if torch.cuda.is_available():
    for index in range(torch.cuda.device_count()):
        props = torch.cuda.get_device_properties(index)
        print(index, props.name, round(props.total_memory / 1024**3, 2), 'GiB')
```
